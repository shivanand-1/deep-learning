```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.optimizers import Adam
from sklearn.datasets import make_moons
from sklearn.model_selection import train_test_split
```

```python
X,y=make_moons(n_samples=1000,noise=0.1)
X.shape
```

```python
model=Sequential()
model.add(Dense(128,activation="relu",input_dim=2))
model.add(Dense(129,activation="softmax"))
model.add(Dense(1,activation="sigmoid"))
```

```python
model.summary()
```

```python
adam=Adam(learning_rate=0.2)
model.compile(loss="binary_crossentropy",optimizer=adam,metrics=["accuracy"])
```

```python
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.4,random_state=64)
histroy=model.fit(X_train,y_train,epochs=1000)
```

```python
y_pred=model.predict(X_test)
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(6,6))

plt.scatter(
    X[:,0],      # x-axis
    X[:,1], 
    c=y,# y-axis
             # color based on class
    cmap="coolwarm",
    edgecolor="k"
)

plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.title("Make Moons Dataset")
plt.show()
```

```python
from mlxtend.plotting import plot_decision_regions
plot_decision_regions(X,y.astype("int"),clf=model,metrics=["accuracy"])
```

```python
plot_decision_regions(
    X,
    y.astype(int),
    clf=model,
    legend=2
)

plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.title("Decision Boundary")
plt.show()
```

```python
plt.plot(histroy.history["loss"])
plt.plot(histroy.history["val_loss"])
```

```python
import tensorflow
from tensorflow.keras.optimizers import Adam
```

```python
model=Sequential()
model.add(Dense(128,activation="relu",input_dim=2,kernel_regularizer=tensorflow.keras.regularizers.L2(0.2)))
model.add(Dense(129,activation="softmax",kernel_regularizer=tensorflow.keras.regularizers.L2(0.2)))
model.add(Dense(1,activation="sigmoid"))
```

```python
model.summary()
```

```python
adam=Adam(learning_rate=0.2)
```

```python
model.compile(loss="binary_crossentropy",optimizer=adam,metrics=["accuracy"])
```

```python
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.4,random_state=64)
histroy=model.fit(X_train,y_train,epochs=1000)
```

```python
plot_decision_regions(
    X,
    y.astype(int),
    clf=model,
    legend=2
)

plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.title("Decision Boundary")
plt.show()
```
