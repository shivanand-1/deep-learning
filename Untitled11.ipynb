```python
import numpy as np

y_train = np.array([
    -0.6561, -0.3099, -0.59035, -0.50855, -0.285,
    -0.2443, -0.02445, 0.00135, -0.2006, 0.07475,
    -0.1422, 0.06515, 0.15265, 0.3521, 0.28415,
    0.5524, 0.23115, 0.20835, 0.4211, 0.60485
])
```

```python
import numpy as np

y_test = np.array([
    -0.69415, -0.451, -0.43005, -0.4484, -0.1475,
    -0.5019, -0.28055, 0.24595, -0.21425, -0.0286,
    0.23415, 0.46575, 0.07955, 0.1973, 0.0719,
    0.3639, 0.5536, 0.3365, 0.50705, 0.33435
])
```

```python
plt.scatter(x_train,y_train,c="black",label="train")
plt.scatter(x_test,y_test,c="yellow",label="train")
```

```python
model=Sequential()
model.add(Dense(128,input_dim=1,activation="relu"))
model.add(Dense(128,activation="relu"))
model.add(Dense(1,activation="linear"))
adam=Adam(learning_rate=0.01)
model.compile(loss="mse",optimizer=adam,metrics=['mse'])
```

```python
model.summary()
```

```python
early_stop = EarlyStopping(
    monitor='val_loss',
    patience=5
)
```

```python
histroy=model.fit(x_train,y_train,epochs=100,validation_data=(x_test, y_test))
```

```python
train_mse=model.evaluate(x_train,y_train,verbose=0)
test_mse=model.evaluate(x_train,y_test,verbose=0)
```

```python
print(f"trian:{train_mse[0]},test_mse{test_mse[0]}")
```

```python
y_pred=model.predict(x_test)
print(y_pred)
```

```python
plt.figure()
plt.title('p=0.2')
plt.scatter(x_train,y_train,c="black",label="train")
plt.scatter(x_test,y_test,c="red",label="test")
plt.plot(x_test,y_pred,color="blue")
plt.legend()
plt.show()
```

```python
plt.plot(histroy.history["loss"])
plt.plot(histroy.history["val_loss"])
plt.show()
```
