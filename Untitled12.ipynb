```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
data=pd.read_csv("ushape.csv")
```

```python
df=pd.DataFrame(data)
```

```python
from sklearn.model_selection import train_test_split
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense,Dropout
from tensorflow.keras.callbacks import EarlyStopping
```

```python
model=Sequential()
```

model.add(optimizer="binary_crossentropy",imput_dim=2)

```python
X=df.iloc[:,0:2]
```

```python
y=df.iloc[:,-1]
```

```python
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.4,random_state=64)
```

```python
early=EarlyStopping(
    monitor='loss'
,patience=4
)
```

```python
history=model.fit(X_train,y_train,epochs=10000,callbacks=[early])
```

```python
model.evaluate(X_train,y_train)
```

```python
plt.plot(history.history["loss"],label="train",color="red")
plt.plot(history.history["accuracy"],label="test")
plt.grid()
plt.legend()
plt.xlabel("epochs")
plt.ylabel("accuacy")
plt.show()
```
