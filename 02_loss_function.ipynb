```python
def perceptron(X,y):
    w1=w2=b=1
    lr=0.1
    for i in range(X.shape[0]):
        z=w1*X[i][0]+w2*X[i][1]+b   
        if z*y[i]>0:
            w1=w1+lr*y[i]*X[i][0]
            w2=w2+lr*y[i]*X[i][1]
            b=b+lr*y[i]
    return w1,w2,b
w1,w2,b=perceptron(X,y)
```

```python
m=-(w1/w2)
c=-(b/w2)
```
