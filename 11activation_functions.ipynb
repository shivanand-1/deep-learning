```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-5, 5, 200)
y = np.tanh(x)

plt.plot(x, y)
plt.grid(True)
plt.title("tanh Activation Function")
plt.xlabel("Input")
plt.ylabel("Output")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

X1=np.linspace(-10,10,1000)
y1=1/(1+np.exp(-X1))
plt.plot(X1,y1)
plt.xlabel("input")
plt.ylabel("output")
plt.title("sigmoid")
plt.show()
```
