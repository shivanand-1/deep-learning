```python
from tensorflow.keras.layers import Conv2D,Flatten,Dense
from tensorflow.keras.datasets import mnist as mt
from sklearn.model_selection import train_test_split
from tensorflow.keras.models import Sequential
```

```python
model=Sequential()
```

```python
model.add(Conv2D(32,activation="relu",input_shape=(124,124,3),padding="same",kernel_size=(3,3)))
model.add(Conv2D(32,activation="relu",padding="same",kernel_size=(3,3)))
model.add(Conv2D(32,activation="relu",padding="same",kernel_size=(3,3)))
model.add(Flatten())
model.add(Dense(128,activation="relu"))
model.add(Dense(32,activation="softmax"))
```
